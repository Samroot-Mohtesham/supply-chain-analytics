# Key Findings

## Delivery Performance

- 54.83% of records were flagged with late-delivery risk.
- 57.28% of records had a positive shipping delay and were classified as Delayed.
- Only 18.70% of records were delivered exactly on time.
- Most delays were between 1 and 2 days.

## Shipping Modes

- First Class had the highest late delivery risk (95.3%).
- Second Class had the second highest late delivery risk (76.6%).
- Standard Class showed the best performance among all shipping methods.

## Market Analysis

- Late delivery risk was consistent across all markets (~54-55%).
- No significant regional delivery problem was identified.

## Department Analysis

- Late delivery risk was similar across product departments.
- Product category was not a major driver of delays.

## Root Cause Hypothesis

- Actual shipping time exceeds scheduled shipping time by an average of 0.57 days.
- Delivery promises appear more aggressive than operational capability.

```python
def delivery_category(delay):
    if delay < 0:
        return "Early"
    elif delay == 0:
        return "On Time"
    else:
        return "Delayed"
```

### Regional Delivery Analysis

- Delivery risk was relatively consistent across regions.
- The highest delay risk was Central Africa (57.96%).
- The lowest delay risk was Canada (48.80%).
- Regional differences were much smaller compared to shipping mode differences.
- Geography does not appear to be the main driver of delivery delays.

### Category Profitability Analysis

- Fishing generated the highest sales ($6.93M) and profit ($756K).
- Cleats generated $4.43M in sales and $495K in profit.
- Camping & Hiking generated $4.12M in sales and $427K in profit.
- Revenue and profit are concentrated within a small number of categories.
- Categories with the highest margins are not necessarily the most profitable.
- Fishing outperformed higher-margin categories due to significantly higher sales volume.
- Computers showed unusually high revenue per order despite a low number of orders.

### Low Performing Categories

- Strength Training generated only $332 profit despite $54.9K in sales.
- Men's Clothing showed a low profit margin of 5.28%.
- Books and Video Games generated relatively low profit compared to sales volume.
- Product category performance varies significantly across the business.
- Category selection appears to have a stronger impact on profitability than customer segment or geographic region.

### Discount Impact Analysis

- Discounted orders represented the majority of transactions.
- Low discount orders generated the highest sales ($16.35M) and profit ($1.88M).
- Increasing discounts reduced average profit margin.
- High discounts did not generate proportional profit growth.
- Low discount strategies appear to provide the best balance between sales volume and profitability.

### Customer Segment and Category Analysis

- Consumer customers are the largest drivers of profitable categories.
- Consumer + Fishing generated the highest profit combination ($393K).
- Fishing performed strongly across all customer segments, indicating broad market demand.
- Corporate customers showed unexpected strength in Women's Apparel, generating nearly $1M in sales and $109K profit.
- Customer segmentation reveals purchasing patterns that are hidden when analysing categories alone.

### Country-Level Profitability Analysis

- USA generated the highest overall sales ($4.88M) and profit ($540K).
- France and Mexico showed strong profitability with above-average margins.
- Germany generated significant revenue but had a lower average margin (11.21%) compared with the company average (12.04%).
- Country-level profitability differences appear to be driven mainly by sales volume rather than large margin differences.
- Further investigation is required into product mix and customer behaviour by market.

### Country and Category Analysis

- Fishing is the strongest global category, generating significant profit across multiple countries.
- USA profitability is driven by a diverse portfolio including Fishing, Camping & Hiking, Cleats, and Women's Apparel.
- Germany shows strong performance in Fishing products despite having a lower overall country margin.
- Women's Apparel demonstrates strong profitability in France, Mexico, and USA markets.
- Product mix appears to be a major driver of country-level profitability.

### Country + Category + Delivery Analysis

- High-value categories are frequently affected by delivery delays.
- USA Fishing orders represent the largest delayed profit exposure.
- Germany Fishing shows strong profitability but significant delayed orders.
- Delay duration is consistent across markets (~1.6 days), suggesting operational process issues rather than isolated geographic problems.
- Supply chain improvements should prioritize high-profit category and market combinations.
